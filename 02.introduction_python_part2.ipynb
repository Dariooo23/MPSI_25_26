### Zad. 1 (1 punkt)

Napisz program, który odwraca cyfry liczby podanej przez użytkownika. Przykładowo, jeśli użytkownik poda 1234, program powinien wypisać 4321. Program nie może używać `str()`, a jedynie operacje matematyczne i pętlę `while`.

```python
liczba = int(input("Podaj liczbę: "))

wynik = 0
while liczba != 0:
    wynik = wynik * 10 + liczba % 10
    liczba //= 10

print(wynik)
```

### Zad. 2 (1 punkt)
Napisz program, który wczyta zdanie od użytkownika i zliczy, ile słów się w nim znajduje.
Za słowo uznajemy ciąg znaków oddzielony spacjami. Pamiętaj, aby pętla mogła obsługiwać różne przypadki (np. dodatkowe spacje).

```python
zdanie = input("Podaj zdanie: ")

liczba_slow = 0
i = 0
n = len(zdanie)

while i < n:
    while i < n and zdanie[i] == ' ':
        i += 1
    if i < n:
        liczba_slow += 1
        while i < n and zdanie[i] != ' ':
            i += 1

print(f"Liczba słów: {liczba_slow}")
```

### Zad. 3 (1 punkt)
Napisz program, który wypisze wszystkie liczby pierwsze od 1 do n, gdzie n wprowadza użytkownik.
Program powinien użyć pętli i instrukcji warunkowej do sprawdzenia, czy dana liczba jest liczbą pierwszą.

```python
n = int(input("Podaj n: "))

sito = [True] * (n + 1)
sito[0] = sito[1] = False

p = 2
while p * p <= n:
    if sito[p]:
        for wielokrotnosc in range(p * p, n + 1, p):
            sito[wielokrotnosc] = False
    p += 1

liczby_pierwsze = [i for i in range(2, n + 1) if sito[i]]
print(liczby_pierwsze)
```

### Zad. 7 (1 punkt) 

1. Zdefiniuj funkcję $ f(x) = \sin(x) $.
2. Zdefiniuj funkcję $ g(x) = x - \frac{x^3}{3!} + \frac{x^5}{5!} $, będącą przybliżeniem $ \sin(x) $ (rozwinięcie Taylora).
3. Narysuj obie funkcje na jednym wykresie w przedziale $ x \in [-2\pi, 2\pi] $.
4. Użyj `matplotlib` do rysowania wykresów, wyróżniając obie funkcje różnymi kolorami i stylami linii.


```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial

def f(x):
    return np.sin(x)

def g(x):
    return x - x**3 / factorial(3) + x**5 / factorial(5)

x = np.linspace(-2 * np.pi, 2 * np.pi, 500)

plt.plot(x, f(x), color="blue",   linestyle="-",  label="f(x) = sin(x)")
plt.plot(x, g(x), color="orange", linestyle="--", label="g(x) = x - x³/3! + x⁵/5!")
plt.xlabel("x")
plt.ylabel("y")
plt.title("sin(x) vs przybliżenie Taylora")
plt.legend()
plt.grid()
plt.show()
```

### Zad. 8 (1 punkt)
Napisz program w Pythonie, który znajduje wszystkie liczby naturalne od 1 do 1000, które spełniają jednocześnie następujące warunki:
- Są liczbami pierwszymi.
- Ich cyfry w systemie dziesiętnym tworzą rosnący ciąg arytmetyczny.
- Liczba cyfr tej liczby jest liczbą pierwszą.
- Suma cyfr tej liczby jest podzielna przez liczbę jej cyfr.
- Program powinien zwrócić listę takich liczb.

```python
def czy_pierwsza(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def czy_rosnacy_ciag_arytmetyczny(n):
    cyfry = []
    while n > 0:
        cyfry.append(n % 10)
        n //= 10
    cyfry.reverse()

    if len(cyfry) < 2:
        return True
    
    roznica = cyfry[1] - cyfry[0]

    if roznica <= 0:
        return False
    
    for i in range(2, len(cyfry)):
        if cyfry[i] - cyfry[i - 1] != roznica:
            return False
        
    return True


def czy_liczba_cyfr_jest_pierwsza(n):
    liczba_cyfr = len(str(n))
    return czy_pierwsza(liczba_cyfr)


def czy_suma_cyfr_podzielna_przez_liczbe_cyfr(n):
    cyfry = []
    while n > 0:
        cyfry.append(n % 10)
        n //= 10
    cyfry.reverse()
    
    return sum(cyfry) % len(cyfry) == 0


wynik = []
for n in range(1, 1001):
    if (czy_pierwsza(n) and
        czy_rosnacy_ciag_arytmetyczny(n) and
        czy_liczba_cyfr_jest_pierwsza(n) and
        czy_suma_cyfr_podzielna_przez_liczbe_cyfr(n)):
        wynik.append(n)

print(wynik)
```

### Zad. 9 (2 punkty)
Zaimplementuj funkcję `tajemnicza_funkcja(n: int) -> int`, która dla dowolnej liczby naturalnej $n$ spełnia następujące warunki:
1. Funkcja zwraca liczbę, która jest iloczynem wszystkich unikalnych dzielników pierwszych liczby $n$.
2. Jeśli $n$ jest liczbą pierwszą, funkcja zwraca $2n+1.$
3. Jeśli $n$ jest potęgą liczby pierwszej, funkcja zwraca $n+$ najmniejszy dzielnik większy od 1.
4. Jeśli $n$ jest liczbą doskonałą, funkcja zwraca sumę cyfr $n$ pomnożoną przez największy dzielnik $n$ mniejszy od $n$.


```python
def jest_pierwsza(n: int) -> bool:
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def jest_potega_pierwszej(n: int) -> bool:
    if n < 4:
        return False
    for p in range(2, int(n ** 0.5) + 1):
        if jest_pierwsza(p) and n % p == 0:
            while n % p == 0:
                n //= p
            return n == 1
    return False


def jest_doskonala(n: int) -> bool:
    if n < 2:
        return False
    suma = 1
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            suma += i
            if i != n // i:
                suma += n // i
    return suma == n


def iloczyn_unikalnych_dzielnikow_pierwszych(n: int) -> int:
    iloczyn = 1
    d = 2
    while d * d <= n:
        if n % d == 0:
            iloczyn *= d
            while n % d == 0:
                n //= d
        d += 1
    if n > 1:
        iloczyn *= n
    return iloczyn


def najmniejszy_dzielnik(n: int) -> int:
    for i in range(2, n + 1):
        if n % i == 0:
            return i
    return n


def najwiekszy_wlasciwy_dzielnik(n: int) -> int:
    for i in range(n // 2, 0, -1):
        if n % i == 0:
            return i
    return 1


def suma_cyfr(n: int) -> int:
    s = 0
    while n > 0:
        s += n % 10
        n //= 10
    return s


def tajemnicza_funkcja(n: int) -> int:
    if jest_pierwsza(n):
        return 2 * n + 1
    elif jest_potega_pierwszej(n):
        return n + najmniejszy_dzielnik(n)
    elif jest_doskonala(n):
        return suma_cyfr(n) * najwiekszy_wlasciwy_dzielnik(n)
    else:
        return iloczyn_unikalnych_dzielnikow_pierwszych(n)


liczba = int(input("Podaj liczbę: "))
wynik = tajemnicza_funkcja(liczba)
print(f"tajemnicza_funkcja({liczba}) = {wynik}")
```
