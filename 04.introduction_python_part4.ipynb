```python
class Person():
    # this method is called constructor
    def __init__(self, name: str, age: int, interests, favourite_animal="dog") -> None: # note that we do not put any type constraints on interest argument
        self.name = name
        self.age = age
        self.interests = interests
        self.favourite_animal = favourite_animal

    def return_age(self):
        return self.age
    
    def introduction(self):
        greeting = "Hello, my name is " + self.name + ". I work at the JU. "
        greeting = greeting + "\t My fav animal is "
        greeting += self.favourite_animal
        greeting += ".\nFrom new line: And I'm " + str(self.return_age()) + ". "
        greeting += f"Again I'm {self.return_age()}" # no need of str here
        return greeting
```

### Zad. 1 (1 punkt) 
Stwórz klasę `Licznik`, która ma atrybut wartość (domyślnie 0) oraz metody `zwieksz()` i `zmniejsz()`, które zwiększają lub zmniejszają wartość o 1. Dodaj metodę `pobierz_wartość()`, zwracającą aktualną wartość.

```python
class Licznik():
    def __init__(self, wartosc: int = 0) -> None:
        self.wartosc = wartosc
    
    def zwieksz(self, ile: int = 1):
        self.wartosc += ile

    def zmniejsz(self, ile: int = 1):
        self.wartosc -= ile
    
    @property
    def pobierz_wartosc(self):
        return self.wartosc
```

### Zad. 2 (3 punkty)

Stwórz prosty model biblioteki.

Zaimplementuj następujące klasy:

- `Ksiazka` – przechowuje podstawowe informacje o książce: `tytul` oraz `autor`.
- `Biblioteka` – zawiera kolekcję książek (np. 52 książki). Powinna mieć metodę umożliwiającą posortowanie książek w kolejności alfabetycznej (wg nazwiska autora).
- `Czytelnik` – posiada listę 5 wypożyczonych książek. Maksymalna liczba wypożyczonych książek to 5.

Następnie napisz metodę sprawdzającą, czy czytelnik ma:

- dwie książki tego samego autora
- trzy książki tego samego autora
- trzy książki jednego autora i dwie książki innego autora

```python
class Ksiazka:
    def __init__(self, tytul: str, autor: str):
        self.tytul = tytul
        self.autor = autor
    
    def __repr__(self):
        return f"Ksiazka('{self.tytul}', '{self.autor}')"


class Biblioteka:
    def __init__(self):
        self.ksiazki = []
    
    def dodaj_ksiazke(self, ksiazka: Ksiazka):
        self.ksiazki.append(ksiazka)
    
    def sortuj_alfabetycznie(self):
        return sorted(self.ksiazki, key=lambda k: k.autor.split()[-1])
    
    def __repr__(self):
        return f"Biblioteka({len(self.ksiazki)} książek)"


class Czytelnik:
    def __init__(self, imie: str):
        self.imie = imie
        self.wypozyczone = []
    
    def wypozycz_ksiazke(self, ksiazka: Ksiazka):
        if len(self.wypozyczone) < 5:
            self.wypozyczone.append(ksiazka)
            return True
        return False
    
    def zwroc_ksiazke(self, ksiazka: Ksiazka):
        if ksiazka in self.wypozyczone:
            self.wypozyczone.remove(ksiazka)
            return True
        return False
    
    def ma_dwie_ksiazki_tego_autora(self, autor: str) -> bool:
        liczba = sum(1 for k in self.wypozyczone if k.autor == autor)
        return liczba >= 2
    
    def ma_trzy_ksiazki_tego_autora(self, autor: str) -> bool:
        liczba = sum(1 for k in self.wypozyczone if k.autor == autor)
        return liczba >= 3
    
    def ma_3_i_2_ksiazki_roznych_autorow(self) -> bool:
        liczby_ksiazek = {}
        for ksiazka in self.wypozyczone:
            liczby_ksiazek[ksiazka.autor] = liczby_ksiazek.get(ksiazka.autor, 0) + 1
        
        # Sprawdzenie czy są dokładnie 3 i 2 książki różnych autorów
        wartosci = sorted(liczby_ksiazek.values(), reverse=True)
        return wartosci == [3, 2]
    
    def __repr__(self):
        return f"Czytelnik('{self.imie}', {len(self.wypozyczone)}/5 książek)"


# Przykład użycia
biblioteka = Biblioteka()

k1 = Ksiazka("Pan Tadeusz", "Adam Mickiewicz")
k2 = Ksiazka("Zbrodnia i kara", "Fiodor Dostojewski")
k3 = Ksiazka("Dziady", "Adam Mickiewicz")
k4 = Ksiazka("Zbrodnia i kara cz. 2", "Fiodor Dostojewski")
k5 = Ksiazka("Ballady", "Adam Mickiewicz")

biblioteka.dodaj_ksiazke(k1)
biblioteka.dodaj_ksiazke(k2)
biblioteka.dodaj_ksiazke(k3)
biblioteka.dodaj_ksiazke(k4)
biblioteka.dodaj_ksiazke(k5)

czytelnik = Czytelnik("Jan")
czytelnik.wypozycz_ksiazke(k1)
czytelnik.wypozycz_ksiazke(k2)
czytelnik.wypozycz_ksiazke(k3)
czytelnik.wypozycz_ksiazke(k4)
czytelnik.wypozycz_ksiazke(k5)

print(f"Czytelnik: {czytelnik}")
print(f"Wypożyczone książki: {czytelnik.wypozyczone}")
print(f"Ma 2 książki Mickiewicza: {czytelnik.ma_dwie_ksiazki_tego_autora('Adam Mickiewicz')}")
print(f"Ma 3 książki Mickiewicza: {czytelnik.ma_trzy_ksiazki_tego_autora('Adam Mickiewicz')}")
print(f"Ma 3 Mickiewicza i 2 Dostojewskiego: {czytelnik.ma_3_i_2_ksiazki_roznych_autorow()}")
```

### Zad. 3 (1 punkt) 
Zaprojektuj klasę `Zadanie` (z priorytetem, nazwą i statusem), a następnie klasę `KolejkaZadan`, która pozwala dodawać, usuwać i pobierać zadanie o najwyższym priorytecie.

```python
class Zadanie:
    def __init__(self, priorytet: int, nazwa: str, status: str = "nowe"):
        self.priorytet = priorytet
        self.nazwa = nazwa
        self.status = status

    def __repr__(self):
        return f"Zadanie(nazwa='{self.nazwa}', priorytet={self.priorytet}, status='{self.status}')"


class KolejkaZadan:
    def __init__(self):
        self.zadania = []

    def dodaj_zadanie(self, zadanie: Zadanie):
        self.zadania.append(zadanie)

    def usun_zadanie(self, nazwa: str) -> bool:
        for i, zadanie in enumerate(self.zadania):
            if zadanie.nazwa == nazwa:
                del self.zadania[i]
                return True
        return False

    def pobierz_najwyzszy_priorytet(self):
        if not self.zadania:
            return None
        return max(self.zadania, key=lambda z: z.priorytet)

    def __repr__(self):
        return f"KolejkaZadan({len(self.zadania)} zadań)"


# Przykład użycia
kolejka = KolejkaZadan()
kolejka.dodaj_zadanie(Zadanie(2, "Napisać raport", "aktywne"))
kolejka.dodaj_zadanie(Zadanie(5, "Naprawić błąd", "aktywne"))
kolejka.dodaj_zadanie(Zadanie(1, "Odpisać na maila", "nowe"))

print(kolejka)
print("Najwyższy priorytet:", kolejka.pobierz_najwyzszy_priorytet())
print("Usunięto 'Napisać raport':", kolejka.usun_zadanie("Napisać raport"))
print("Po usunięciu:", kolejka.zadania)
```

### Zad. 4 (1 punkt) 
Zaimplementuj klasę `PionekSzachowy` z atrybutami koloru i pozycji (np. "E2"). Dodaj metodę `mozliwe_ruchy()`, zwracającą listę dostępnych ruchów zgodnych z zasadami szachów.

```python
class PionekSzachowy:
    def __init__(self, kolor: str, pozycja: str):
        kolor = kolor.lower()
        if kolor not in ("bialy", "czarny"):
            raise ValueError("Kolor musi być 'bialy' albo 'czarny'.")
        if len(pozycja) != 2 or pozycja[0].upper() not in "ABCDEFGH" or pozycja[1] not in "12345678":
            raise ValueError("Pozycja musi być w formacie np. 'E2'.")

        self.kolor = kolor
        self.pozycja = pozycja.upper()

    def mozliwe_ruchy(self):
        kolumna = self.pozycja[0]
        wiersz = int(self.pozycja[1])
        ruchy = []

        if self.kolor == "bialy":
            if wiersz < 8:
                ruchy.append(f"{kolumna}{wiersz + 1}")
            if wiersz == 2:
                ruchy.append(f"{kolumna}{wiersz + 2}")
        else:
            if wiersz > 1:
                ruchy.append(f"{kolumna}{wiersz - 1}")
            if wiersz == 7:
                ruchy.append(f"{kolumna}{wiersz - 2}")

        return ruchy

    def __repr__(self):
        return f"PionekSzachowy(kolor='{self.kolor}', pozycja='{self.pozycja}')"


# Przykład użycia
pionek_bialy = PionekSzachowy("bialy", "E2")
pionek_czarny = PionekSzachowy("czarny", "D7")

print(pionek_bialy, "->", pionek_bialy.mozliwe_ruchy())
print(pionek_czarny, "->", pionek_czarny.mozliwe_ruchy())
```

### Zad. 5 (1 punkt)
Napisz klasę `Proces` (id, użycie CPU, stan – aktywne, zatrzymane, zakończone). Następnie klasę `MenadzerProcesow`, który pozwala na tworzenie, usuwanie i monitorowanie procesów.

```python
class Proces:
    def __init__(self, pid: int, cpu_uzycie: float, stan: str = "aktywne"):
        dozwolone_stany = {"aktywne", "zatrzymane", "zakonczone"}
        if stan not in dozwolone_stany:
            raise ValueError("Stan musi być jednym z: aktywne, zatrzymane, zakonczone")
        if cpu_uzycie < 0:
            raise ValueError("Użycie CPU nie może być ujemne")

        self.pid = pid
        self.cpu_uzycie = cpu_uzycie
        self.stan = stan

    def __repr__(self):
        return f"Proces(pid={self.pid}, cpu={self.cpu_uzycie}%, stan='{self.stan}')"


class MenadzerProcesow:
    def __init__(self):
        self.procesy = {}

    def utworz_proces(self, proces: Proces) -> bool:
        if proces.pid in self.procesy:
            return False
        self.procesy[proces.pid] = proces
        return True

    def usun_proces(self, pid: int) -> bool:
        if pid in self.procesy:
            del self.procesy[pid]
            return True
        return False

    def monitoruj_procesy(self):
        return list(self.procesy.values())

    def pobierz_proces(self, pid: int):
        return self.procesy.get(pid)


# Przykład użycia
menadzer = MenadzerProcesow()
menadzer.utworz_proces(Proces(101, 23.5, "aktywne"))
menadzer.utworz_proces(Proces(102, 0.0, "zatrzymane"))
menadzer.utworz_proces(Proces(103, 88.1, "aktywne"))

print("Wszystkie procesy:", menadzer.monitoruj_procesy())
print("Proces 102:", menadzer.pobierz_proces(102))
print("Usunięto 101:", menadzer.usun_proces(101))
print("Po usunięciu:", menadzer.monitoruj_procesy())
```

Dana jest klasa `Produkt`, która przechowuje informacje o nazwie i cenie produktu. Obecnie oblicza cenę końcową z podatkiem VAT (23%).

1. Zmodyfikuj metodę `oblicz_cene_z_vat()`, aby obsługiwała różne stawki VAT – stawka powinna być podawana jako argument.
2. Dodaj nową metodę `rabat()`, która obniży cenę o podany procent.

```python
class Produkt:
    def __init__(self, nazwa: str, cena: float):
        self.nazwa = nazwa
        self.cena = cena

    def oblicz_cene_z_vat(self, stawka_vat: float):
        return self.cena * (1 + stawka_vat / 100)

    def rabat(self, procent: float):
        self.cena *= (1 - procent / 100)
        return self.cena
```

### Zad. 7 (2 punkty)

Napisz klasę `Point`, która reprezentuje punkt w przestrzeni 2D o współrzędnych `(x, y)`. 
Punkty traktujemy jak **wektory zaczepione w początku układu współrzędnych**.

Uzupełnij brakujące metody w klasie.

Klasa powinna obsługiwać:

#### Operacje na wektorach
- `__add__(self, other)` – dodawanie wektorów
- `__sub__(self, other)` – odejmowanie wektorów
- `__mul__(self, scalar)` – mnożenie wektora przez skalar

#### Operacje geometryczne
- `length(self)` – długość wektora
- `distance(self, other)` – odległość między punktami
- `dot(self, other)` – iloczyn skalarny
- `angle(self, other)` – kąt między wektorami (w radianach)

#### Operacje dodatkowe
- `normalize(self)` – zwraca wektor jednostkowy
- `rotate(self, angle)` – obraca wektor o zadany kąt (w radianach)
- `__repr__(self)` – czytelna reprezentacja punktu

```python
import math

class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Point(self.x * scalar, self.y * scalar)

    def length(self):
        return math.sqrt(self.x ** 2 + self.y ** 2)

    def distance(self, other):
        return (self - other).length()

    def dot(self, other):
        return self.x * other.x + self.y * other.y

    def angle(self, other):
        dl_self = self.length()
        dl_other = other.length()
        if dl_self == 0 or dl_other == 0:
            raise ValueError("Nie można liczyć kąta z wektorem zerowym.")

        cos_kat = self.dot(other) / (dl_self * dl_other)
        cos_kat = max(-1.0, min(1.0, cos_kat))
        return math.acos(cos_kat)

    def normalize(self):
        dl = self.length()
        if dl == 0:
            raise ValueError("Nie można normalizować wektora zerowego.")
        return Point(self.x / dl, self.y / dl)

    def rotate(self, angle):
        cos_a = math.cos(angle)
        sin_a = math.sin(angle)
        x_new = self.x * cos_a - self.y * sin_a
        y_new = self.x * sin_a + self.y * cos_a
        return Point(x_new, y_new)

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"
```
