### 3. Mutowalne i niemutowalne typy danych

W Pythonie typy danych można podzielić na **mutowalne** (mutable) i **niemutowalne** (immutable).  
- **Mutowalne** – ich zawartość można zmieniać po utworzeniu. Przykłady: `list`, `dict`, `set`.
- **Niemutowalne** – raz utworzone, nie mogą być zmienione. Przykłady: `int`, `float`, `str`, `tuple`.


````python
# Przykład 1: liczby są niemutowalne
x = 10
print(id(x))  # ID obiektu przed zmianą
x += 5
print(id(x))  # ID obiektu po zmianie (jest inne)
````

````python
# Przykład 2: stringi są niemutowalne
s = "Hello"
print(id(s))
s += " World"
print(id(s))  # Nowy obiekt
````

````python
# Przykład 3: krotki (tuple) są niemutowalne
t = (1, 2, 3)
print(id(t))
t += (4,)  # Tworzy nową krotkę
print(id(t))  # Inny obiekt
````

````python
# Przykład 4: listy są mutowalne
lst = [1, 2, 3]
print(id(lst))  # ID przed zmianą
lst.append(4)
print(id(lst))  # ID po zmianie (takie samo)
````

````python
# Przykład 5: słowniki (dict) są mutowalne
d = {"a": 1, "b": 2}
print(id(d))
d["c"] = 3
print(id(d))  # ID pozostaje to samo
````

````python
# Przykład 6: zbiory (set) są mutowalne
s = {1, 2, 3}
print(id(s))
s.add(4)
print(id(s))  # ID się nie zmienia
````

### Zad. 7 (1 pkt)
Stwórz krotkę dni_tygodnia zawierającą nazwy dni tygodnia. Spróbuj zmienić "czwartek" na "czwartek(tłusty czwartek)". Co się stanie?
Jak można zmodyfikować tę krotkę, skoro jest niemutowalna?

````python
dni_tygodnia = ("poniedziałek", "wtorek", "środa", "czwartek", "piątek", "sobota", "niedziela")
print("Krotka dni tygodnia:", dni_tygodnia)

# Próba zmiany elementu krotki – krotki są niemutowalne, to wywoła błąd:
try:
    dni_tygodnia[3] = "czwartek(tłusty czwartek)"
except TypeError as e:
    print(f"\nBłąd: {e}")
    print("Krotki są niemutowalne – nie można zmieniać ich elementów!")

# Jak obejść niemutowalność? Konwertujemy na listę, modyfikujemy, tworzymy nową krotkę:
lista_dni = list(dni_tygodnia)
lista_dni[3] = "czwartek(tłusty czwartek)"
dni_tygodnia = tuple(lista_dni)
print("\nZmodyfikowana krotka:", dni_tygodnia)
````

### Zad. 8 (3 pkt)
#### Symulacja rankingu piosenek

W serwisie muzycznym użytkownik tworzy ranking swoich ulubionych utworów.

- **Top 2 ulubione piosenki** zapisane są jako **krotka** (niemutowalna – nie można jej zmieniać).
- **Nowe propozycje tygodnia** zapisane są jako **lista** (mutowalna – można ją modyfikować).

Każda piosenka zapisana jest w postaci:

(tytuł, liczba_punktów)

Następnie:
1. Dodaj do listy `propozycje` nową piosenkę:  
`("Dance The Night", 92)`
2. Usuń z listy `propozycje` piosenkę z najmniejszą liczbą punktów. Możesz użyć funkcji `min()` z parametrem `key`.
3. Połącz krotkę `ulubione` oraz listę `propozycje` w jedną nową listę zawierającą wszystkie piosenki. Nie modyfikuj krotki `ulubione`.
4. Posortuj wszystkie piosenki malejąco według liczby punktów. Użyj parametru `key` w funkcji `sort()` lub `sorted()`.
5. Wybierz trzy najwyżej ocenione piosenki (TOP 3).
6. Wypisz:
    - ulubione piosenki  
    - aktualne propozycje  
    - pełny ranking (po sortowaniu)  
    - TOP 3 tygodnia  

Zadbaj o czytelne formatowanie wyników.

#### Dane początkowe

```python
ulubione = (("Blinding Lights", 95), ("Shape of You", 90))

propozycje = [("Flowers", 85), ("As It Was", 88), ("Levitating", 87)]

````python
ulubione = (("Blinding Lights", 95), ("Shape of You", 90))

propozycje = [("Flowers", 85), ("As It Was", 88), ("Levitating", 87)]

# 1. Dodaj nową piosenkę do propozycji
propozycje.append(("Dance The Night", 92))

# 2. Usuń piosenkę z najmniejszą liczbą punktów
min_piosenka = min(propozycje, key=lambda p: p[1])
propozycje.remove(min_piosenka)

# 3. Połącz krotkę ulubione i listę propozycje
wszystkie = list(ulubione) + propozycje

# 4. Posortuj malejąco według liczby punktów
wszystkie.sort(key=lambda p: p[1], reverse=True)

# 5. TOP 3
top3 = wszystkie[:3]

# 6. Wypisz wyniki
print("=" * 40)
print("Ulubione piosenki (Top 2 - niezmienne):")
for tytul, punkty in ulubione:
    print(f"  {tytul} – {punkty} pkt")

print("\nAktualne propozycje tygodnia:")
for tytul, punkty in propozycje:
    print(f"  {tytul} – {punkty} pkt")

print("\nPełny ranking (posortowany malejąco):")
for i, (tytul, punkty) in enumerate(wszystkie, 1):
    print(f"  {i}. {tytul} – {punkty} pkt")

print("\nTOP 3 tygodnia:")
for i, (tytul, punkty) in enumerate(top3, 1):
    print(f"  {i}. {tytul} – {punkty} pkt")
print("=" * 40)
````
