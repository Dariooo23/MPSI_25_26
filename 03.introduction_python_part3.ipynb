### Zad. 1 (1 punkt)
Napisz funkcję, która oblicza sumę liczb od 1 do $n$ na dwa różne sposoby.
Jakie są inne sposoby obliczenia takiej sumy? 
Która z zaimplementowanych metod jest najszybsza? Która zużywa najmniej pamięci? Jakie są zalety i wady każdego podejścia?
Jak upewnić się, że funkcja zwraca poprawne wyniki?

```python
def suma_petla(n):
    suma = 0
    for i in range(1, n + 1):
        suma += i
    return suma

def suma_formula(n):
    return n * (n + 1) // 2

# Inne sposoby obliczania sumy liczb od 1 do n:
# 1. sum(range(1, n+1))  – wbudowana funkcja sum z generatorem
# 2. numpy: np.sum(np.arange(1, n+1))
# 3. funkcja rekurencyjna: suma_rekurencyjna(n) = n + suma_rekurencyjna(n-1)

# Która metoda jest najszybsza?
# Metoda z wzorem Gaussa (suma_formula) – złożoność O(1), tylko jedno działanie arytmetyczne.
# Pętla i sum(range(...)) są O(n) – muszą iterować przez wszystkie elementy.
# numpy może być szybki dla dużych n dzięki wektoryzacji, ale ma narzut inicjalizacji tablicy.

# Która metoda zużywa najmniej pamięci?
# Wzór Gaussa (suma_formula) – przechowuje tylko stałą liczbę zmiennych, O(1) pamięci.
# Pętla for – O(1) pamięci (akumulator), nie tworzy listy.
# sum(range(...)) – range jest generatorem w Python 3, więc też O(1).
# numpy – tworzy tablicę n elementów w pamięci, O(n).
# Rekurencja – O(n) na stosie wywołań, ryzyko przepełnienia dla dużych n.

# Zalety i wady podejść:
# Pętla for:        prosta, czytelna, O(n) czas, O(1) pamięć; wolna dla bardzo dużych n
# Wzór Gaussa:      najszybsza O(1), O(1) pamięć; wymaga znajomości wzoru, tylko dla całkowitych
# sum(range(...)):  pythoniczne, czytelne, O(n) czas, O(1) pamięć
# numpy:            wygodne dla większych obliczeń, ale narzut importu i alokacji tablicy
# Rekurencja:       elegancka, ale O(n) stos, wolna, ryzyko RecursionError

# Jak upewnić się, że funkcja zwraca poprawne wyniki?
# Weryfikacja przez porównanie obu implementacji dla tych samych danych wejściowych.
# Testy jednostkowe (np. za pomocą assert lub modułu unittest/pytest):
#   assert suma_petla(0) == 0
#   assert suma_petla(1) == 1
#   assert suma_petla(10) == 55
#   assert suma_petla(100) == 5050
#   assert suma_petla(n) == suma_formula(n) for n in range(1, 1000)

n = int(input("Podaj n: "))
print("Suma za pomocą pętli:", suma_petla(n))
print("Suma za pomocą formuły:", suma_formula(n))

# Weryfikacja poprawności
assert suma_petla(n) == suma_formula(n), "Wyniki różnią się!"
print("Wyniki zgodne – funkcje działają poprawnie.")
```

### Zad. 2 (1 punkt)
Napisz funkcję, która dla podanych trzech parametrów: $n$ (numer elementu ciągu), $a_1$ (wartość pierwszego elementu ciągu - domyślnie 1), $q$ (wartość ilorazu ciągu geometrycznego - domyślnie 2) zwróci $n$-ty element ciągu geometrycznego.

```python
def nty_element_ciagu_geometrycznego(n, a1, q = 2):
    return a1 * (q ** (n - 1))
```

### Zad. 3 (2 punkty) 
Napisz funkcję `oblicz_pole(figura, *wymiary)`, która przyjmuje nazwę figury ("kwadrat", "prostokąt", "koło") oraz jej wymiary. Funkcja powinna zwracać pole danej figury.
Jakie wymiary są wymagane dla każdej figury?
Co jeśli użytkownik poda zbyt wiele lub za mało parametrów?
Czy dopuszczamy wartości ujemne? Jak wtedy powinna zachować się funkcja?

Jak zmodyfikować tą funkcję, aby zwracała róznież obwód figury, jeśli użytkownik tego zażąda?

```python
import math


def oblicz_pole(figura, *wymiary, zwroc_obwod=False):
    figura = figura.lower()

    if figura == "kwadrat":
        if len(wymiary) != 1:
            raise ValueError("Kwadrat wymaga dokładnie 1 wymiaru: bok.")
        a = wymiary[0]
        if a <= 0:
            raise ValueError("Wymiar musi być dodatni.")
        pole = a ** 2
        obwod = 4 * a

    elif figura == "prostokąt":
        if len(wymiary) != 2:
            raise ValueError("Prostokąt wymaga dokładnie 2 wymiarów: a, b.")
        a, b = wymiary
        if a <= 0 or b <= 0:
            raise ValueError("Wymiary muszą być dodatnie.")
        pole = a * b
        obwod = 2 * (a + b)

    elif figura == "koło":
        if len(wymiary) != 1:
            raise ValueError("Koło wymaga dokładnie 1 wymiaru: promień.")
        r = wymiary[0]
        if r <= 0:
            raise ValueError("Promień musi być dodatni.")
        pole = math.pi * r ** 2
        obwod = 2 * math.pi * r

    else:
        raise ValueError("Nieznana figura. Użyj: 'kwadrat', 'prostokąt', 'koło'.")

    if zwroc_obwod:
        return pole, obwod
    return pole


# Przykłady
print(oblicz_pole("kwadrat", 4))
print(oblicz_pole("prostokąt", 3, 5))
print(oblicz_pole("koło", 2, zwroc_obwod=True))
```

### Zad. 4 (1 punkt) 
Napisz funkcję `czy_liczba_pierwsza(n)`, która sprawdza, czy liczba $n$ jest liczbą pierwszą. Funkcja powinna zwracać `True`, jeśli liczba jest pierwsza, lub `False` w przeciwnym przypadku.

```python
def czy_liczba_pierwsza(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True


n = int(input("Podaj liczbę: "))
if czy_liczba_pierwsza(n):
    print(f"{n} jest liczbą pierwszą.")
else:
    print(f"{n} nie jest liczbą pierwszą.")
```

### Zad. 5 (1 punkt)
Napisz funkcję, która przyjmuje listę liczb i zwraca wszystkie możliwe permutacje tej listy. Nie używaj itertools.permutations.

```python
def permutacje(lista):
    if len(lista) <= 1:
        return [lista[:]]

    wynik = []
    for i in range(len(lista)):
        element = lista[i]
        pozostale = lista[:i] + lista[i + 1:]

        for p in permutacje(pozostale):
            wynik.append([element] + p)

    return wynik


# Przykład
liczby = [1, 2, 3]
wszystkie = permutacje(liczby)
print("Permutacje:")
for p in wszystkie:
    print(p)

print("Liczba permutacji:", len(wszystkie))
```

### Zad 6 (3 punkty).
Napisz funkcję, która oblicza wyznacznik kwadratowej macierzy (determinant) przy użyciu:
1. rozwinięcia Laplace’a,
2. eliminacji Gaussa (rozkład macierzy na iloczyn macierzy trójkątnej dolnej i trójkątnej górnej),
3. funkcji `det`.
Przedstaw porównanie metod pod względem złożoności czasowej i wyników dla przykładowych macierzy rozmiarów: $2 \times 2$ i $200 \times 200$. 

```python
import numpy as np
import time


def det_laplace(macierz):
    """Wyznacznik przez rozwinięcie Laplace'a (rekurencyjnie)."""
    A = np.array(macierz, dtype=float)
    n = A.shape[0]

    if A.shape[0] != A.shape[1]:
        raise ValueError("Macierz musi być kwadratowa.")

    if n == 1:
        return A[0, 0]
    if n == 2:
        return A[0, 0] * A[1, 1] - A[0, 1] * A[1, 0]

    det = 0.0
    for j in range(n):
        minor = np.delete(np.delete(A, 0, axis=0), j, axis=1)
        det += ((-1) ** j) * A[0, j] * det_laplace(minor)
    return det


def det_gauss(macierz):
    """Wyznacznik przez eliminację Gaussa (z częściowym wyborem elementu głównego)."""
    A = np.array(macierz, dtype=float).copy()

    if A.shape[0] != A.shape[1]:
        raise ValueError("Macierz musi być kwadratowa.")

    n = A.shape[0]
    zamiany_wierszy = 0

    for i in range(n):
        pivot = i + np.argmax(np.abs(A[i:, i]))

        if np.isclose(A[pivot, i], 0.0):
            return 0.0

        if pivot != i:
            A[[i, pivot]] = A[[pivot, i]]
            zamiany_wierszy += 1

        for k in range(i + 1, n):
            wsp = A[k, i] / A[i, i]
            A[k, i:] -= wsp * A[i, i:]

    det = np.prod(np.diag(A))
    if zamiany_wierszy % 2 == 1:
        det *= -1

    return float(det)


def det_numpy(macierz):
    """Wyznacznik przez gotową funkcję biblioteczną."""
    A = np.array(macierz, dtype=float)
    if A.shape[0] != A.shape[1]:
        raise ValueError("Macierz musi być kwadratowa.")
    return float(np.linalg.det(A))


def zmierz_czas(funkcja, macierz):
    start = time.perf_counter()
    wynik = funkcja(macierz)
    koniec = time.perf_counter()
    return wynik, koniec - start


# ----- Porównanie dla 2x2 -----
A2 = np.array([[1, 2], [3, 4]], dtype=float)

lap2, t_lap2 = zmierz_czas(det_laplace, A2)
gauss2, t_gauss2 = zmierz_czas(det_gauss, A2)
np2, t_np2 = zmierz_czas(det_numpy, A2)

print("=== Macierz 2x2 ===")
print("Laplace:", lap2, "czas:", f"{t_lap2:.6f}s")
print("Gauss:  ", gauss2, "czas:", f"{t_gauss2:.6f}s")
print("NumPy:  ", np2, "czas:", f"{t_np2:.6f}s")


# ----- Porównanie dla 200x200 -----
np.random.seed(42)
A200 = np.random.randn(200, 200)

print("\n=== Macierz 200x200 ===")
print("Laplace: pominięto (złożoność ~ O(n!), praktycznie niewykonalne dla 200x200)")

gauss200, t_gauss200 = zmierz_czas(det_gauss, A200)
np200, t_np200 = zmierz_czas(det_numpy, A200)

print("Gauss:  ", gauss200, "czas:", f"{t_gauss200:.6f}s")
print("NumPy:  ", np200, "czas:", f"{t_np200:.6f}s")

print("\nRóżnica |Gauss - NumPy| dla 200x200:", abs(gauss200 - np200))


# Wnioski teoretyczne (skrót):
# 1) Laplace: dokładny i dydaktyczny, ale bardzo wolny: O(n!).
# 2) Gauss: praktyczny algorytm ogólny: O(n^3).
# 3) NumPy det: również oparty na szybkich metodach macierzowych (LU), zwykle najszybszy w praktyce.
```

### Zad 7 (1 punkt)
Napisz funkcję, która zwróci wszystkie podzbiory danego zbioru. 

```python
def podzbiory(zbior):
    wynik = [[]]

    for element in zbior:
        nowe = []
        for podzbior in wynik:
            nowe.append(podzbior + [element])
        wynik.extend(nowe)

    return wynik


# Przykład
zbior = [1, 2, 3]
wynik = podzbiory(zbior)

print("Wszystkie podzbiory:")
for p in wynik:
    print(p)

print("Liczba podzbiorów:", len(wynik))
```
